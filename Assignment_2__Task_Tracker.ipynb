# Assignment 2: Task Tracker Using Python

**Objective:**

To develop a simple Task Tracker application using Python that allows users to add, view, complete, and delete tasks.

**Main Features:**
- Add new tasks.
- View all tasks.
- Mark tasks as completed.
- Delete tasks.
- Save tasks in a JSON file.
- Load previously saved tasks.
- Handle errors using exception handling.

**Tools:** Python and Google Colab.

```python
# Initialize an empty task list

tasks = []

print("Task Tracker Initialized Successfully!")
```

```python
# Creating sample tasks

tasks = [
    {
        "title": "Study Python",
        "done": False
    },
    {
        "title": "Complete Assignment",
        "done": False
    }
]

print(tasks)
```

```python
# Function to add a new task

def add_task():
    title = input("Enter your task: ")

    task = {
        "title": title,
        "done": False
    }

    tasks.append(task)

    print("Task added successfully!")
```

```python
add_task()
```

```python
# Function to display all tasks

def view_tasks():

    if not tasks:
        print("No tasks available!")

    else:
        print("\n----- MY TASKS -----")

        for index, task in enumerate(tasks, start=1):

            status = "Completed" if task["done"] else "Pending"

            print(index, ".", task["title"], "-", status)
```

```python
view_tasks()
```

```python
# Function to mark a task as completed

def complete_task():

    view_tasks()

    if not tasks:
        return

    try:
        task_number = int(input("Enter task number to complete: "))

        if 1 <= task_number <= len(tasks):

            tasks[task_number - 1]["done"] = True

            print("Task marked as completed successfully!")

        else:
            print("Invalid task number!")

    except ValueError:
        print("Please enter a valid number!")
```

```python
complete_task()
```

```python
# Function to delete a task

def delete_task():

    view_tasks()

    if not tasks:
        return

    try:
        task_number = int(input("Enter task number to delete: "))

        if 1 <= task_number <= len(tasks):

            removed_task = tasks.pop(task_number - 1)

            print("Task deleted successfully:", removed_task["title"])

        else:
            print("Invalid task number!")

    except ValueError:
        print("Please enter a valid number!")
```

```python
delete_task()
```

```python
import json

print("JSON module imported successfully!")
```

```python
# Function to save tasks in a JSON file

def save_tasks():

    try:
        with open("tasks.json", "w") as file:

            json.dump(tasks, file, indent=4)

        print("Tasks saved successfully!")

    except Exception as e:
        print("Error while saving tasks:", e)
```

```python
save_tasks()
```

```python
# Function to load tasks from a JSON file

def load_tasks():

    global tasks

    try:
        with open("tasks.json", "r") as file:

            tasks = json.load(file)

        print("Tasks loaded successfully!")

    except FileNotFoundError:
        tasks = []
        print("No saved tasks found. Starting with an empty list.")

    except json.JSONDecodeError:
        tasks = []
        print("The JSON file is empty or contains invalid data.")
```

```python
load_tasks()
```

```python
view_tasks()
```

```python
# Main menu of Task Tracker

def main():

    load_tasks()

    while True:

        print("\n===== TASK TRACKER =====")
        print("1. Add Task")
        print("2. View Tasks")
        print("3. Complete Task")
        print("4. Delete Task")
        print("5. Save Tasks")
        print("6. Load Tasks")
        print("7. Exit")

        try:
            choice = int(input("Enter your choice: "))

            if choice == 1:
                add_task()

            elif choice == 2:
                view_tasks()

            elif choice == 3:
                complete_task()

            elif choice == 4:
                delete_task()

            elif choice == 5:
                save_tasks()

            elif choice == 6:
                load_tasks()

            elif choice == 7:
                print("Thank you for using Task Tracker!")
                break

            else:
                print("Invalid choice! Please select 1-7.")

        except ValueError:
            print("Please enter a valid number!")
```

```python
main()
```
